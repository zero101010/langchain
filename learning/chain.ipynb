```python
from  langchain_core.prompts import ChatPromptTemplate

from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser

from dotenv import load_dotenv

from langchain_core.runnables import RunnableLambda
load_dotenv()
```

```python
def UpperCase(text: str) -> str:
    return text.upper()
```

```python
# Create the structure with template and the input expected from user
prompt = ChatPromptTemplate.from_template("Explique sobre {assunto} em 2 linhas.")
# Define the model params
model = init_chat_model(model="gemini-2.5-flash", model_provider="google_genai")
# Define the output parser
parser = StrOutputParser()

final_step = RunnableLambda(UpperCase)
```

```python
chain = prompt | model | parser | final_step
result = chain.invoke({"assunto": "IA"})
print(result)
```
